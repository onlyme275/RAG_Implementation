```python
simple_text = {
    "data/text_files/python_intro.txt": """
    Python is a high-level, general-purpose programming language that is widely used in 
    web development, automation, data science, machine learning, artificial intelligence, 
    cybersecurity, and scripting. Python was designed with an emphasis on readability and 
    simplicity, which makes it popular among beginners as well as experienced developers. 
    Python uses indentation to define blocks of code instead of curly braces, and it supports 
    multiple programming paradigms including procedural, object-oriented, and functional 
    programming. Python is dynamically typed, meaning that variables do not need to be 
    explicitly declared with a data type before they are used. The language provides many 
    built-in data types such as integers, floating-point numbers, strings, lists, tuples, 
    dictionaries, sets, and booleans. Python also provides a large standard library and a 
    huge ecosystem of third-party packages. In data science, libraries such as NumPy, pandas, 
    Matplotlib, and scikit-learn are commonly used for numerical computation, data analysis, 
    visualization, and machine learning. In deep learning, frameworks such as TensorFlow and 
    PyTorch are commonly used to build and train neural networks. Python is also widely used 
    in artificial intelligence and natural language processing because libraries and frameworks
    make it easier to work with text, embeddings, language models, and retrieval systems. 
    A Python program can be divided into modules and functions, allowing developers to 
    organize large applications into smaller reusable components. Functions can accept 
    parameters and return values, while classes can be used to implement object-oriented 
    designs. Python also provides exception handling through try, except, else, and finally
    blocks, allowing programs to handle errors without immediately terminating. Because 
    Python is easy to read, has extensive documentation, and provides libraries for many 
    different areas of computing, it has become one of the most commonly used programming 
    languages for software development, data science, machine learning, and artificial 
    intelligence.
    """
}

for filepath, content in simple_text.items():
    with open(filepath, 'w' , encoding="utf-8") as f:
        f.write(content)
print("writting files sucessfully!")
```

```python
loader = TextLoader("data/text_files/python_intro.txt", encoding="utf-8")
document = loader.load()
print(document)
```
